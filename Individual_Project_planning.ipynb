**2. Questions:**

Broad Question that will be addressed - What player characteristics and behaviours are most predictive of subscribing to a game-related newsletter, and how do these features differ between various player types?

Specific formulated question - Can a player's experience level, playing time, and age predict whether a player subscribes to the game-related newsletter? 

All variables used are on the player level, which will directly show their engagement with the game. Since the newsletter is game-related, player's engagement (in terms of their experience, played hours, and age) will help us to predict/determine whether players will subscribe to the newsletter. 
The data wrangling stage will include handling missing age values and re-coding categorical values for data modelling.

**3. Data Wrangling Stage:**

```r
#Demonstrating dataset loading
players <- read_csv("players.csv")

head(players)
```

```r
#selecting necessary variables

players_select <- players |>
    select(subscribe, experience, played_hours, Age)

#converting categorical variables to factors

players_select <- players_select |>
    mutate(subscribe = as.factor(subscribe), experience = as.factor(experience))

#mean values for numeric variables

mean_summary <- players_select |>
    summarise(mean_played_hours = mean(played_hours, na.rm = TRUE), mean_age = mean(Age, na.rm = TRUE)) |>
    mutate(across(everything(), ~round(.x, 2)))

head(players_select)
mean_summary
```

```r
#Plot 1 - subscription rate by player experience 

players_select |>
    group_by(experience, subscribe) |>
    summarise(count = n(), .groups = "drop") |>
    group_by(experience) |>
    mutate(prop = count / sum(count)) |>
    filter(subscribe == "TRUE") |>
    ggplot(aes(x = experience, y = prop, fill = experience)) +
    scale_y_continuous(labels = scales::percent_format()) +
    geom_col(show.legend = FALSE) + 
    labs(title = "Newsletter Subscription Rate by Experience level", x = "Experience Level", y = "Subscription Rate (%)") + 
    theme(text = element_text(size = 15)) 
    
    
```

```r
#Plot 2 - subscription rate by age
options(repr.plot.width = 14, repr.plot.height = 7)

ggplot(players_select, aes(x = factor(Age), fill = subscribe)) +
    geom_bar(color = "black", position = "stack") +
    labs(title = "Newsletter Subscription Rate by Player's Age", x = "Age (years)", y = "Count", fill = "Subscribed") +
    theme(text = element_text(size = 18), axis.text.x = element_text(angle = 90, vjust = 0.5, hjust = 1)
  )  
```

**Methods and Plan:**

- **Proposed Method -** k-Nearest Neighbors (k-NN) Classification

Using k-NN would be an appropriate method to address our question of interest becuase it is flexible and non-parametric. That means that it doesn't make any underlying assumptions about data distribution. This is important because our data includes categorical (experience) and continuous (age and played_hours) variables. k-NN typically does not have many assumptions; however, it requires features to be appropriately scaled or normalized as this ensures proper distance calculations. It also assumes that points close to each other in space are more likely to have the same class label. When it comes to weaknesses associated with the k-NN model, it slows considerably as the training data increases, it may not perform as well with a larger number of predictors, and it is very sensitive to irrelevant or noisy features that may degrade its performance. The choice of 'k' is also important as too small of a 'k' leads to more variance and too high of a 'k' produces more bias. When comparing and selecting the model, metrics of accuracy, precision, and recall will be used. Additionally, cross-validation (ex. 5-fold) will be performed on the training data to tune 'k' and assess model stability. To apply the k-NN model, a 75% training and 25% testing split will be used before modelling to avoid leakage. One initial split into training and test sets will be done and further splitting will occur within the training set during cross-validation. Lastly, a k-fold cross-validation (5-fold) will be used on the training data.
